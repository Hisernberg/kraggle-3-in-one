## How the score works

Each task has an official verifier. It runs the circuit on many sampled inputs, checks every answer exactly, and bills a cost:

| Task | Cost | The letters |
|---|---|---|
| invsqrt | $Q \cdot C$ | $Q$ qubits; $C = N_{\mathrm{Tof}} + N_T / 4$, the mean Toffolis and T gates of one shot |
| DQI | $R \cdot Q^2 \cdot T$ | $T$ the mean Toffolis of one shot; $R \ge 1$ grows with the failed shots |
| Chemistry | $\lambda_{\mathrm{eff}} \cdot T \cdot Q$ | $\lambda_{\mathrm{eff}}$ the Hamiltonian's effective normalization; $T$ the Toffolis of one step |

A task's **ratio** is the organizers' reference cost divided by this cost. The references are 2,076,774, $2.4439 \times 10^{13}$ and 3,355,599,588. The score is the **geometric mean of the three ratios**:

$$\text{score} = \left( \frac{C^{\mathrm{ref}}_{\mathrm{invsqrt}}}{C_{\mathrm{invsqrt}}} \cdot \frac{C^{\mathrm{ref}}_{\mathrm{DQI}}}{C_{\mathrm{DQI}}} \cdot \frac{C^{\mathrm{ref}}_{\mathrm{chem}}}{C_{\mathrm{chem}}} \right)^{1/3}$$

So making any one task 1% cheaper lifts the score by about 0.33%.

The three bills reward different things. In DQI, $Q$ is squared: near 880 qubits one qubit is worth 0.23% of the cost. In invsqrt one qubit near 63 is worth 1.6%. In chemistry each of the three factors counts the same.

## Chemistry · our team's R8 B143 factorization with full-XOR five-word angle lookups

**The task.** Block-encode the FeMoCo active-space Hamiltonian, 76 orbitals and 113 electrons, for qubitized phase estimation. The submission supplies a factorization of the Hamiltonian (double factorization with tensor hypercontraction, DF-THC) and the circuits that load its rotation angles. Two error measures of the factorization must stay below 20 mHa, and the sum-of-squares gap $E_{\mathrm{gap}}$ must stay above 3.914898 Ha.

The bill is **$\lambda_{\mathrm{eff}} \times T \times Q$**. $\lambda_{\mathrm{eff}}$ is the normalization after spectrum amplification (the "SOS" in the task's name), and it sets how many walk steps phase estimation needs. $T$ is the Toffolis of one step and $Q$ its qubits.

**This submission.**
- $\lambda_{\mathrm{eff}}$ **48.56091**, $T$ **16,528**, $Q$ **430**.
- Errors 19.999 and 19.999 mHa, against the 20 mHa limit. Gap 3.914902 Ha, against 3.914898.
- Cost **345,124,297**, 9.7229× below the reference. That is 0.75% below our previous version, at 347,715,360, and 1.21% below the best public design of others, heliosli's, with $\lambda_{\mathrm{eff}}$ 47.4940, $T$ 17,066 and $Q$ 431. Ours pays 2.2% in $\lambda_{\mathrm{eff}}$ for 3.2% fewer Toffolis and one qubit less.
- Unlike those two, every angle circuit here is the XOR oracle of the task statement on any words in its target registers, not only on the words the verifier gives it. That costs 64 Toffolis a step, 0.39%.

**The factorization.** Eight groups ($R = 8$) of 143 vectors ($B = 143$), rank $C = 60$, every angle stored as an 8-bit word:
- The team pruned 11 vectors from each group of our $B = 154$ design by exact compensated deletion: each removal refits the remaining vectors so that the represented Hamiltonian moves as little as possible. There are now 76 + 8 × 143 = 1,220 addresses instead of 1,308, and 143 vectors fill nine 16-way blocks of the inner state preparation where 154 need ten: 5,463 Toffolis instead of 5,981, at rank 62. Rank 60 needs 5,311.
- Circuit $k$ loads the XOR of the words of batches $k$ and $k - 1$. In 11 of the 15 circuits, the lowest bit of each of the five words it loads is, on every block of 16 consecutive addresses, an affine function of the address bits (a first-order Reed–Muller code word), so it is almost free to load. In three of them, circuits 9, 11 and 13, the bit above it is coded the same way. All the words were quantized to 8 bits jointly with these codes.
- The fit aimed at a slightly biased target: a small positive coherent term keeps the gap above its floor, where a pure error fit would leave it just below. Both errors are always measured against the true Hamiltonian.
- With every word fixed, the weights and offsets were fitted for the lowest $\lambda_{\mathrm{eff}}$ within the error and gap limits. That took $\lambda_{\mathrm{eff}}$ from 49.15808 to 48.56091 and used up the slack of both errors.

**The angle lookups.** Each lookup circuit adds to the five 8-bit words in its target registers the five words of the address it is given. Each output bit is an XOR of products of address bits. One Toffoli computes the AND of two XOR combinations, so it covers two products at once.
- No qubit of the words controls a Toffoli or serves as workspace. Each product's Toffoli targets one of the word bits that the product changes, between two fans of CNOTs from that bit to the others, so the product is added to all of them whatever they held. Each circuit is therefore the oracle $|a\rangle|z\rangle \to |a\rangle|z \oplus \delta_k(a)\rangle$ of the task statement for every $z$. The best public designs and our previous versions clear each word's top bit with a CNOT and use the five freed qubits as workspace, which is exact only on words whose top two bits are equal: those are the words the family streams from zero, which is what the verifier feeds it when every earlier circuit has loaded its words exactly.
- Without those five qubits, a circuit within 128 qubits keeps 8 products of the low address bits cached at once instead of 11, so the coded circuits compute some of them twice: 2 or 3 more Toffolis each, 32 a pass.
- The four circuits without coded bits plan their 225 products in phases over a cache of low-address products, after heliosli's method: 305 Toffolis each, at 126 qubits.
- The eight with one coded bit per word need 203 products: 286 Toffolis each (one needs 285), at 128 qubits. The three with two coded bits per word need 181: 264 Toffolis each.
- A pass costs 4,299 Toffolis, against 4,680 in heliosli's design.
